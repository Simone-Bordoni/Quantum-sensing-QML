## Imports and Setup

```python
import numpy as np
import jax
import jax.numpy as jnp
from qsopt import *

print(f"JAX backend: {jax.devices()}")
```

## Creating Trainable Angle Parameters

```python
from qsopt.core.trainable_parameters import ParameterConstraints, ParameterGroup

# create rotation angles group
rotation_constraints = ParameterConstraints(periodic=True, period=2*np.pi)
rotation_angles = ParameterGroup(
    name = "rotation_angles", 
    param_type=ParameterType.ROTATION_ANGLE, 
    initial_values=[np.pi/4, np.pi/6], 
    constraints=rotation_constraints)

print(rotation_angles)

# Create trainable parameters container
trainable_parameters = TrainableParameters()
trainable_parameters.add_parameter_group(rotation_angles)

print(trainable_parameters)

# It is also possible to create parameter groups directly when adding them to the container
trainable_parameters = TrainableParameters()
trainable_parameters.add_parameter_group(
    "rotation_angles",
    param_type=ParameterType.ROTATION_ANGLE,
    initial_values=[np.pi/4, np.pi/6],
    constraints=ParameterConstraints(periodic=True, period=2*np.pi)
)

print(trainable_parameters)

# Finally it is possible to directly add angle parameters
trainable_parameters = TrainableParameters()
trainable_parameters.add_rotation_angles("rotation_angles", [np.pi/4, np.pi/6])

print(trainable_parameters)

# Add a different type of parameter group, e.g., measurement times
time_constraints = ParameterConstraints(min_value=0.0)
trainable_parameters.add_parameter_group(
    "measurement_time",
    param_type=ParameterType.MEASUREMENT_TIME,
    initial_values=[1.0],
    constraints=time_constraints
)

print(trainable_parameters)
```

## Working with Angle Parameters

Access, modify, and work with the angle parameters we've defined.

```python
# Get all parameters as a flat array (useful for optimization)
all_params = trainable_parameters.get_parameter_vector()
print(f"All parameters as array: {all_params}")

# Update individual parameter groups
trainable_parameters["rotation_angles"].update_values(jnp.array([np.pi/2, np.pi/3]))

print(trainable_parameters)

# Test bounds enforcement by applying constraints
test_params = jnp.array([-np.pi, 4*np.pi])
print(f"Before constraint application:\n {test_params}")
# Set these parameters (constraints will be applied automatically)
trainable_parameters.set_parameter_vector(test_params)

# Check the constrained values
constrained_params = trainable_parameters.get_parameter_vector()
print(f"After constraint application (periodicity):\n {constrained_params}")
```
